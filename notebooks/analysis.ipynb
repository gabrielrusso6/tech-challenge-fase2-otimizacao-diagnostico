```python
# Carregar e explorar os dados
df = pd.read_csv('../data/breast-cancer-wisconsin.csv')

# Visualizar as primeiras linhas
print("Shape do dataset:", df.shape)
print("\nPrimeiras linhas:")
display(df.head())

# Informações sobre o dataset
print("\nInformações do dataset:")
df.info()

# Estatísticas descritivas
print("\nEstatísticas descritivas:")
display(df.describe())

# Verificar valores ausentes
print("\nValores ausentes por coluna:")
print(df.isnull().sum())

# Distribuição da variável alvo
print("\nDistribuição da variável alvo (diagnosis):")
print(df['diagnosis'].value_counts())
print("\nProporção:")
print(df['diagnosis'].value_counts(normalize=True))
```

```python
# Visualizações exploratórias
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração dos gráficos
plt.style.use('seaborn-v0_8')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. Distribuição da variável alvo
sns.countplot(data=df, x='diagnosis', ax=axes[0, 0])
axes[0, 0].set_title('Distribuição de Diagnóstico')
axes[0, 0].set_xlabel('Diagnóstico (M=Malígno, B=Benigno)')

# 2. Histograma de radius_mean
sns.histplot(data=df, x='radius_mean', hue='diagnosis', kde=True, ax=axes[0, 1])
axes[0, 1].set_title('Distribuição do Raio Médio por Diagnóstico')

# 3. Scatter plot: radius_mean vs texture_mean
sns.scatterplot(data=df, x='radius_mean', y='texture_mean', hue='diagnosis', ax=axes[1, 0])
axes[1, 1].set_title('Raio Médio vs Textura Média')

# 4. Boxplot de area_mean por diagnóstico
sns.boxplot(data=df, x='diagnosis', y='area_mean', ax=axes[1, 1])
axes[1, 1].set_title('Área Média por Diagnóstico')

plt.tight_layout()
plt.show()

# Matriz de correlação das features principais
features_corr = ['radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 
                 'smoothness_mean', 'compactness_mean', 'concavity_mean', 
                 'concave points_mean', 'symmetry_mean', 'fractal_dimension_mean']

plt.figure(figsize=(12, 10))
correlation_matrix = df[features_corr].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, fmt='.2f')
plt.title('Matriz de Correlação das Features Médias')
plt.tight_layout()
plt.show()
```

```python
# Análise estatística avançada
from scipy import stats

# Separar features numéricas (excluindo id e diagnosis)
numeric_features = df.select_dtypes(include=[np.number]).columns
numeric_features = [col for col in numeric_features if col != 'id']

# Teste estatístico para diferença entre benigno e maligno
print("Teste t de Student para diferença entre grupos:")
print("-" * 50)

benigno = df[df['diagnosis'] == 'B']
maligno = df[df['diagnosis'] == 'M']

significant_features = []
for feature in numeric_features[:10]:  # Analisar primeiras 10 features
    t_stat, p_value = stats.ttest_ind(benigno[feature], maligno[feature])
    print(f"{feature}: p-value = {p_value:.6f}")
    if p_value < 0.05:
        significant_features.append(feature)
        print(f"  -> Significativo (p < 0.05)")
    else:
        print(f"  -> Não significativo")
    print()

print(f"Features significativamente diferentes: {len(significant_features)}")
print(significant_features)

# Estatísticas por grupo
print("\nEstatísticas por grupo para features principais:")
print("-" * 50)
key_features = ['radius_mean', 'texture_mean', 'area_mean', 'concavity_mean']

for feature in key_features:
    print(f"\n{feature}:")
    print(f"  Benigno:  média={benigno[feature].mean():.2f}, desv={benigno[feature].std():.2f}")
    print(f"  Maligno:  média={maligno[feature].mean():.2f}, desv={maligno[feature].std():.2f}")
```
